## 1. Why resistance evolves

HIV replicates rapidly and generates genetic variation. When antiretroviral therapy fully suppresses replication, the virus has little opportunity to evolve. When replication continues in the presence of a drug, variants that replicate better under that selective pressure can become enriched.

A resistance mutation does not appear because HIV “tries” to escape a drug. Mutations arise during replication; drug pressure changes which variants are favored. ## 3. How to read mutation names

A mutation such as **M184V** is compact amino-acid notation: **M** is the reference amino acid (methionine), **184** is the protein position, and **V** is the observed amino acid (valine). So M184V means *methionine at position 184 has changed to valine*.

Other familiar examples include **K103N** (NNRTI-associated), **Q148H** (INSTI-associated pathway), and **L90M** (PI-associated). A mutation is not equivalent to the same resistance effect for every drug; drug-specific effects and combinations matter. Stanford therefore uses mutation-specific comments and penalty scores rather than a single binary mutation flag.

Source: [Stanford HIVDB release notes](https://hivdb.stanford.edu/page/release-notes/).

```python
amino_acids = {
    "A":"alanine", "C":"cysteine", "D":"aspartic acid", "E":"glutamic acid",
    "F":"phenylalanine", "G":"glycine", "H":"histidine", "I":"isoleucine",
    "K":"lysine", "L":"leucine", "M":"methionine", "N":"asparagine",
    "P":"proline", "Q":"glutamine", "R":"arginine", "S":"serine",
    "T":"threonine", "V":"valine", "W":"tryptophan", "Y":"tyrosine"
}

def explain_mutation(mutation):
    old, new, position = mutation[0], mutation[-1], mutation[1:-1]
    return f"{mutation}: {amino_acids.get(old, old)} → {amino_acids.get(new, new)} at amino-acid position {position}."

for mutation in ["M184V", "K103N", "Q148H", "L90M"]:
    print(explain_mutation(mutation))
```

## 4. Acquired, transmitted, and pretreatment resistance

WHO distinguishes three related concepts:

- **Acquired HIV drug resistance (ADR):** resistance emerges while virus is replicating during antiretroviral exposure.
- **Transmitted HIV drug resistance (TDR):** a person acquires HIV that already contains resistance mutations.
- **Pretreatment HIV drug resistance (PDR):** resistance is detected before starting or restarting ART; it can reflect transmitted resistance, previous drug exposure, or both.

## 5. Stanford HIVdb: resistance is graded, not simply yes/no

Stanford HIVdb sums drug-specific mutation penalty scores and maps the total to five interpretation levels:

| HIVdb score | Interpretation |
|---:|---|
| **0–9** | Susceptible |
| **10–14** | Potential low-level resistance |
| **15–29** | Low-level resistance |
| **30–59** | Intermediate resistance |
| **≥60** | High-level resistance |

This matters because the same genotype may preserve substantial activity for one drug while reducing activity of another drug in the same class. The evidence base for a drug-resistance mutation can include selection during therapy, in-vitro susceptibility, clinical virologic response, and whether the mutation is a common natural polymorphism.

Source: [Stanford HIVDB release notes](https://hivdb.stanford.edu/page/release-notes/).

At a high level, HIVdb accepts protease/reverse-transcriptase/integrase sequence or mutations, performs quality assessment, identifies relevant mutations, applies drug-specific rules/penalty scores, and returns resistance levels with comments. Clinicians must then integrate those results with treatment history and other patient-specific information.

**Key boundary:** HIVdb is a genotype-interpretation engine, not an autonomous regimen recommender. Source: [Stanford HIVDB release notes](https://hivdb.stanford.edu/page/release-notes/).

## 7. Clinical interpretation vs. population surveillance

| Clinical genotype interpretation | Population surveillance |
|---|---|
| asks what resistance an individual's virus appears to have | asks how resistance is distributed across populations, places, or periods |
| uses an individual genotype plus clinical context | aggregates many sequences or studies |
| HIVdb scores/comments can be central | SDRM/CPR definitions, sampling, and epidemiologic denominators are central |
| goal: support treatment interpretation | goal: monitor public-health patterns |

Stanford exposes both the **HIVdb Program** and surveillance resources such as the **Transmitted Resistance Surveillance Map / CPR**, reflecting these distinct purposes. This distinction is the bridge to the rest of this repository.

### Three misconceptions to avoid
1. **“A mutation means the drug is useless.”** No: effects are drug-specific and context-dependent.
2. **“Resistance only happens because someone misses doses.”** Adherence matters, but resistance also reflects replication, pharmacology, regimen potency, prior exposure, transmitted resistance, and genetic barrier.
3. **“A country-colored resistance map is a national treatment-failure rate.”** No: surveillance maps inherit the sampling frame, study design, timing, resistance definitions, and denominator limitations of their source data.
